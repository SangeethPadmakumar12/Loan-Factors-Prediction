1. DATASET OVERVIEW

```python
print("1. BEFORE CLEANING SHAPE:", A.shape)
print("\nMissing Values:\n", A.isnull().sum())
print("\nDuplicate Rows:\n", A.duplicated().sum())
```

2. MISSING VALUES HANDLING

Numeric - Median

```python
for col in ['AnnualIncome','LoanAmountRequested','CreditScore','ExistingLoansCount','LatePaymentsLastYear', 'Age']:
  if col in A.columns:
    A[col] = A[col].fillna(A[col].median())
```

Categorical - Mode

```python
for col in ['Gender','MaritalStatus','EducationLevel','EmploymentStatus','PurposeOfLoan','LoanApproved']:
  if col in A.columns:
    A[col] = A[col].fillna(A[col].mode()[0])
```

```python
print("2. After Fill - Total Missing:", A.isnull().sum().sum())
```

3. INCONSISTENCY FIXING

A. Gender

```python
if 'Gender' in A.columns:
  A['Gender'] = A['Gender'].astype(str).str.strip().str.title()
```

B. MaritalStatus

```python
if 'MaritalStatus' in A.columns:
  A['MaritalStatus'] = A['MaritalStatus'].astype(str).str.strip().str.title()
```

C. PurposeOfLoan

```python
if 'PurposeOfLoan' in A.columns:
  A['PurposeOfLoan'] = A['PurposeOfLoan'].astype(str).str.strip().str.title()
```

D. EducationalLevel

```python
if 'EducationLevel' in A.columns:
  A['EducationLevel'] = A['EducationLevel'].astype(str).str.strip().str.title()
```

E. EmploymentStatus

```python
if 'EmploymentStatus' in A.columns:
  A['EmploymentStatus'] = A['EmploymentStatus'].astype(str).str.strip().str.title()
```

F. LoanApproved

```python
if 'LoanApproved' in A.columns:
  A['LoanApproved'] = A['LoanApproved'].astype(str).str.strip().str.title()
```

4.Duplicates

```python
before = A.shape[0]
A = A.drop_duplicates()
```

```python
A = A.drop_duplicates(subset=['CustomerID'], keep='first')
print(f"4. Duplicates Removed: {before - A.shape[0]} | New Shape: {A.shape}")
```

5. OUTLIERS (IQR METHOD)

```python
for col in ['AnnualIncome', 'LoanAmountRequested', 'CreditScore']:
  if col in A.columns:
    Q1 = A[col].quantile(0.25)
    Q3 = A[col].quantile(0.75)
    IQR = Q3 - Q1
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
    count = ((A[col] < lower) | (A[col] > upper)).sum()
    print(f"{col}: {count} outliers | Actual Min:{A[col].min()} Max:{A[col].max()}")

```

```python
print(f"\nCreditScore valid range is 300-850")
print(f"Your data range: {A['CreditScore'].min()} to {A['CreditScore'].max()}")
```

6. RENAME COLUMNS

```python
A.rename(columns={
'AnnualIncome': 'annual_income',
'LoanAmountRequested': 'loan_amt',
'PurposeOfLoan': 'loan_purpose',
'CreditScore': 'credit_score',
'ExistingLoansCount': 'existing_loans',
'LatePaymentsLastYear': 'late_payments',
'LoanApproved': 'loan_approved',
'MaritalStatus': 'marital_status',
'EducationLevel': 'education',
'EmploymentStatus': 'employment'
}, inplace=True)
```

```python
print("6. Renamed Columns:", A.columns.tolist())
```

7. NEW COLUMNS

A. Loan to Income Ratio

```python
A['loan_to_income_ratio'] = A['loan_amt'] / A['annual_income']
```

B. Credit Score Category

```python
A['credit_category'] = pd.cut(A['credit_score'],
bins=[0, 580, 670, 740, 800, 1000],
labels=['Poor', 'Fair', 'Good', 'Very Good', 'Excellent'])
```

C. Income Category

```python
income_max = A['annual_income'].max()
A['income_category'] = pd.cut(A['annual_income'],
bins=[0, 40000, 70000, 100000, income_max],
labels=['Low','Medium','High','Very High'])
```

D. Risk Factor

```python
A['is_high_risk'] = 0
A.loc[A['credit_score'] < 600, 'is_high_risk'] = 1
A.loc[A['loan_to_income_ratio'] > 0.4, 'is_high_risk'] = 1
A.loc[A['late_payments'] > 2, 'is_high_risk'] = 1
A.loc[A['existing_loans'] >= 3, 'is_high_risk'] = 1
```

E. Total Loan Burden

```python
A['total_burden'] = A['existing_loans'] + A['late_payments']
```

```python
print("\n7. FINAL SHAPE:", A.shape)
print(A[['loan_to_income_ratio', 'credit_category', 'income_category', 'is_high_risk', 'total_burden']].head())
```
